# 01 - Data Loading & Initial Inspection

## Overview
This notebook downloads the **IBM Telco Customer Churn** dataset (if not already cached locally in `data/`), loads it into a Pandas DataFrame, and performs initial inspection.

**Dataset Source URL:** `https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv`

```python
import os
import urllib.request
import pandas as pd
import numpy as np

# Ensure data path
data_dir = os.path.join("..", "data") if os.path.exists(os.path.join("..", "data")) else "data"
os.makedirs(data_dir, exist_ok=True)
data_path = os.path.join(data_dir, "Telco-Customer-Churn.csv")
url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"

# Download raw dataset if not present locally
if not os.path.exists(data_path):
    print(f"Downloading raw dataset from {url}...")
    urllib.request.urlretrieve(url, data_path)
    print("Download complete.")
else:
    print(f"Raw CSV dataset already exists at {data_path}.")

# Load raw CSV into DataFrame
df = pd.read_csv(data_path)
print("Dataset loaded successfully.")
```

```python
# 1. Dataset Shape
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"Exact Shape Tuple: {df.shape}")
```

```python
# 2. Data Types per Column
print("--- Data Types ---")
print(df.dtypes)
```

```python
# 3. First 5 Rows
df.head()
```

```python
# 4. Missing Values Count per Column
print("--- Standard Missing Values (df.isnull().sum()) ---")
print(df.isnull().sum())

print("\n--- Blank String Count in TotalCharges ---")
blank_total_charges = (df['TotalCharges'].str.strip() == '').sum()
print(f"Blank space (' ') entries in 'TotalCharges': {blank_total_charges}")
```

```python
# 5. Class Balance of Churn Column
churn_counts = df['Churn'].value_counts()
churn_percentages = df['Churn'].value_counts(normalize=True) * 100

churn_summary = pd.DataFrame({
    'Count': churn_counts,
    'Percentage (%)': churn_percentages.round(2)
})
print("--- Churn Class Balance ---")
print(churn_summary)
```

## Key Findings & Inspection Notes

- **Dataset Dimensions**: Contains **7,043 rows** and **21 columns**.
- **Data Type Issue (`TotalCharges`)**: `TotalCharges` is stored as an `object` (string) dtype instead of numeric (float64).
- **Blank Values**: Standard `df.isnull().sum()` reports `0` missing values across all columns. However, `TotalCharges` contains **11 blank space values** (`" "`).
- **Class Balance (`Churn`)**: 5,174 "No" (73.46%) and 1,869 "Yes" (26.54%). The churn rate is roughly **26.5%**, showing moderate class imbalance.
